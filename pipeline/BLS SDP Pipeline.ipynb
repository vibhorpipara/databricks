```python
#Imports
from pyspark import pipelines as dp
from pyspark.sql.functions import col,concat_ws,trim
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    IntegerType,
    DoubleType,
    ArrayType
)
```

```python

# Bronze tables for BLS

# Bronze: BLS All Data
# Source: pr.data.1.AllData

@dp.materialized_view(
    name="bronze_bls_all_data",
    comment="BLS historical productivity data from pr.data.1.AllData"
)
@dp.expect("valid_series_id", "series_id IS NOT NULL")
@dp.expect("valid_year", "year IS NOT NULL")
@dp.expect("valid_period", "period IS NOT NULL")
@dp.expect("valid_value", "value IS NOT NULL")
def bronze_bls_all_data():
    all_data_schema = StructType([
    StructField("series_id", StringType(), True),
    StructField("year", IntegerType(), True),
    StructField("period", StringType(), True),
    StructField("value", DoubleType(), True),
    StructField("footnote_codes", StringType(), True)
    ])
    return (
        spark.read
        .format("csv")
        .option("header", True)
        .option("sep", "\t")
        .schema(all_data_schema)
        .load("/Volumes/bls/bronze/bls_raw/raw/pr.data.1.AllData")
    )

# Bronze: series data
# Source: pr.series     


@dp.materialized_view(
    name="bronze_bls_series",
    comment="BLS series metadata from pr.series"
)
@dp.expect("valid_series_id", "series_id IS NOT NULL")
@dp.expect("valid_sector_code", "sector_code IS NOT NULL")
@dp.expect("valid_class_code", "class_code IS NOT NULL")
@dp.expect("valid_measure_code", "measure_code IS NOT NULL")
def bronze_bls_series():
    series_schema = StructType([
    StructField("series_id", StringType(), True),
    StructField("sector_code", IntegerType(), True),
    StructField("class_code", IntegerType(), True),
    StructField("measure_code", IntegerType(), True),
    StructField("duration_code", IntegerType(), True),
    StructField("seasonal", StringType(), True),
    StructField("base_year", StringType(), True),
    StructField("footnote_codes", StringType(), True),
    StructField("begin_year", IntegerType(), True),
    StructField("begin_period", StringType(), True),
    StructField("end_year", IntegerType(), True),
    StructField("end_period", StringType(), True)
    ])
    return (
        spark.read
        .format("csv")
        .option("header", True)
        .option("sep", "\t")
        .schema(series_schema)
        .load("/Volumes/bls/bronze/bls_raw/raw/pr.series")
    )

# Bronze: sector data
# Source: pr.sector

@dp.materialized_view(
    name="bronze_bls_sector",
    comment="BLS sector metadata from pr.sector"
)
@dp.expect("valid_sector_code", "sector_code IS NOT NULL")
@dp.expect("valid_sector_name", "sector_name IS NOT NULL")
def bronze_bls_sector():
    sector_schema = StructType([
        StructField("sector_code", IntegerType(), True),
        StructField("sector_name", StringType(), True),
        StructField("display_level", IntegerType(), True),
        StructField("selectable", StringType(), True),
        StructField("sort_sequence", IntegerType(), True)
    ])

    return (
        spark.read
        .format("csv")
        .option("header", True)
        .option("sep", "\t")
        .schema(sector_schema)
        .load("/Volumes/bls/bronze/bls_raw/raw/pr.sector")
    )

# Bronze: class data
# Source: pr.class

@dp.materialized_view(
    name="bronze_bls_class",
    comment="BLS class metadata from pr.class"
)
@dp.expect("valid_class_code", "class_code IS NOT NULL")
@dp.expect("valid_class_text", "class_text IS NOT NULL")
def bronze_bls_class():
    class_schema = StructType([
        StructField("class_code", IntegerType(), True),
        StructField("class_text", StringType(), True),
        StructField("display_level", IntegerType(), True),
        StructField("selectable", StringType(), True),
        StructField("sort_sequence", IntegerType(), True)
    ])

    return (
        spark.read
        .format("csv")
        .option("header", True)
        .option("sep", "\t")
        .schema(class_schema)
        .load("/Volumes/bls/bronze/bls_raw/raw/pr.class")
    )
# Bronze: measure data
# Source: pr.measure

@dp.materialized_view(
    name="bronze_bls_measure",
    comment="BLS measure metadata from pr.measure"
)
@dp.expect("valid_measure_code", "measure_code IS NOT NULL")
@dp.expect("valid_measure_text", "measure_text IS NOT NULL")
def bronze_bls_measure():
    measure_schema = StructType([
        StructField("measure_code", IntegerType(), True),
        StructField("measure_text", StringType(), True),
        StructField("display_level", IntegerType(), True),
        StructField("selectable", StringType(), True),
        StructField("sort_sequence", IntegerType(), True)
    ])

    return (
        spark.read
        .format("csv")
        .option("header", True)
        .option("sep", "\t")
        .schema(measure_schema)
        .load("/Volumes/bls/bronze/bls_raw/raw/pr.measure")
    )


# Bronze: duration data
# Source: pr.duration

@dp.materialized_view(
    name="bronze_bls_duration",
    comment="BLS duration metadata from pr.duration"
)
@dp.expect("valid_duration_code", "duration_code IS NOT NULL")
@dp.expect("valid_duration_text", "duration_text IS NOT NULL")
def bronze_bls_duration():
    duration_schema = StructType([
        StructField("duration_code", IntegerType(), True),
        StructField("duration_text", StringType(), True),
        StructField("display_level", IntegerType(), True),
        StructField("selectable", StringType(), True),
        StructField("sort_sequence", IntegerType(), True)
    ])

    return (
        spark.read
        .format("csv")
        .option("header", True)
        .option("sep", "\t")
        .schema(duration_schema)
        .load("/Volumes/bls/bronze/bls_raw/raw/pr.duration")
    )


# Bronze: seasonal data
# Source: pr.seasonal

@dp.materialized_view(
    name="bronze_bls_seasonal",
    comment="BLS seasonal metadata from pr.seasonal"
)
@dp.expect("valid_seasonal_code", "Seasonal_code IS NOT NULL")
@dp.expect("valid_seasonal_text", "Seasonal_text IS NOT NULL")
def bronze_bls_seasonal():
    seasonal_schema = StructType([
        StructField("Seasonal_code", StringType(), True),
        StructField("Seasonal_text", StringType(), True)
    ])

    return (
        spark.read
        .format("csv")
        .option("header", True)
        .option("sep", "\t")
        .schema(seasonal_schema)
        .load("/Volumes/bls/bronze/bls_raw/raw/pr.seasonal")
    )    
```

```python
# Bronze tables for population

# Bronze: Population data
# Source: DataUSA population API

@dp.materialized_view(
    name="bronze_population",
    comment="US population data from DataUSA API"
)
@dp.expect("valid_year", "year IS NOT NULL")
@dp.expect("valid_population", "population IS NOT NULL")
def bronze_population():

    population_record_schema = StructType([
        StructField("Nation ID", StringType(), True),
        StructField("Nation", StringType(), True),
        StructField("Year", IntegerType(), True),
        StructField("Population", DoubleType(), True)
    ])

    population_schema = StructType([
        StructField(
            "data",
            ArrayType(population_record_schema),
            True
        )
    ])

    return (
        spark.read
        .format("json")
        .option("multiLine", True)
        .schema(population_schema)
        .load("/Volumes/bls/bronze/bls_raw/raw/datausa/population.json")
        .selectExpr("explode(data) AS record")
        .select(
            col("record.Nation ID").alias("Nation_ID"),
            col("record.Nation").alias("Nation"),
            col("record.Year").alias("year"),
            col("record.Population").alias("population")
        )
    )
```

```python
# Silver Tables for population

# Silver: Population
# Source: bronze_population

@dp.materialized_view(
    name="bls.silver.silver_population",
    comment="Cleaned US population data from Bronze"
)
@dp.expect("valid_year", "year IS NOT NULL")
@dp.expect("valid_population", "population IS NOT NULL")
def silver_population():
    return (
        spark.read.table("bls.bronze.bronze_population")
        .select(
            "Nation_ID",
            "Nation",
            "year",
            "population"
        )
        .dropDuplicates(["Nation_ID", "year"])
    )
```

```python
# Silver Table for BLS all data - Combines BLS values with series metadata and lookup descriptions

# Silver: BLS enriched data


@dp.materialized_view(
    name="bls.silver.silver_bls_data",
    comment="Enriched BLS data with human-readable series metadata"
)
@dp.expect("valid_series_id", "series_id IS NOT NULL")
@dp.expect("valid_year", "year IS NOT NULL")
@dp.expect("valid_period", "period IS NOT NULL")
@dp.expect("valid_value", "value IS NOT NULL")
def silver_bls_data():

    all_data = (
        spark.read.table("bls.bronze.bronze_bls_all_data")
        .withColumn("series_id", trim(col("series_id")))
    )

    series = (
        spark.read.table("bls.bronze.bronze_bls_series")
        .withColumn("series_id", trim(col("series_id")))
    )

    sector = spark.read.table("bls.bronze.bronze_bls_sector")
    class_data = spark.read.table("bls.bronze.bronze_bls_class")
    measure = spark.read.table("bls.bronze.bronze_bls_measure")
    duration = spark.read.table("bls.bronze.bronze_bls_duration")
    seasonal = spark.read.table("bls.bronze.bronze_bls_seasonal")

    return (
        all_data.alias("d")
        .join(series.alias("s"), "series_id", "left")
        .join(sector.alias("sec"), "sector_code", "left")
        .join(class_data.alias("cls"), "class_code", "left")
        .join(measure.alias("m"), "measure_code", "left")
        .join(
            duration.alias("dur"),
            "duration_code",
            "left"
        )
        .join(
            seasonal.alias("sea"),
            col("sea.Seasonal_code") == col("s.seasonal"),
            "left"
        )
        .select(
            "d.series_id",
            "d.year",
            "d.period",
            "d.value",
            "d.footnote_codes",
            "s.sector_code",
            "sec.sector_name",
            "s.class_code",
            "cls.class_text",
            "s.measure_code",
            "m.measure_text",
            "s.duration_code",
            "dur.duration_text",
            "s.seasonal",
            "sea.Seasonal_text",
            "s.base_year",
            "s.begin_year",
            "s.begin_period",
            "s.end_year",
            "s.end_period",
            concat_ws(
                " | ",
                col("sec.sector_name"),
                col("cls.class_text"),
                col("m.measure_text"),
                col("dur.duration_text"),
                col("sea.Seasonal_text")
            ).alias("series_label")
        )
    )
```

```python
# Gold Layers
# question 1 - Population mean & standard deviation
# Spark SQL


@dp.materialized_view(
    name="bls.gold.gold_population_statistics",
    comment="Mean and standard deviation of annual US population from 2013 through 2018"
)
def gold_population_statistics():
    return spark.sql("""
        SELECT
            AVG(population) AS mean_population,
            STDDEV_POP(population) AS population_stddev
        FROM bls.silver.silver_population
        WHERE year BETWEEN 2013 AND 2018
    """)


# Alternative PySpark implementation:
#
# from pyspark.sql.functions import avg, stddev_pop
#
# def gold_population_statistics_pyspark():
#     return (
#         spark.read.table("bls.silver.silver_population")
#         .filter("year BETWEEN 2013 AND 2018")
#         .agg(
#             avg("population").alias("mean_population"),
#             stddev_pop("population").alias("population_stddev")
#         )
#     )
```

```python
# Gold Layers
# question 2 - Best year for each series_id
# Spark SQL

@dp.materialized_view(
    name="bls.gold.gold_bls_best_year",
    comment="Best year for each BLS series based on the largest annual sum of quarterly values"
)
def gold_bls_best_year():

    return spark.sql("""
        WITH annual_values AS (
            SELECT
                series_id,
                series_label,
                year,
                SUM(value) AS annual_value
            FROM bls.silver.silver_bls_data
            WHERE period IN ('Q01', 'Q02', 'Q03', 'Q04') 
            GROUP BY
                series_id,
                series_label,
                year
        ),

        ranked_years AS (
            SELECT
                series_id,
                series_label,
                year,
                annual_value,
                ROW_NUMBER() OVER (
                    PARTITION BY series_id
                    ORDER BY annual_value DESC, year DESC
                ) AS rn
            FROM annual_values
        )

        SELECT
            series_id,
            series_label,
            year AS best_year,
            annual_value
        FROM ranked_years
        WHERE rn = 1
    """)


# ============================================================
# Alternative implementation - PySpark
# ============================================================

# from pyspark.sql.functions import col, row_number, sum
# from pyspark.sql.window import Window

# annual_values = (
#     spark.read.table("bls.silver.silver_bls_data")
#     .filter(
#         col("period").isin("Q01", "Q02", "Q03", "Q04")
#     )
#     .groupBy(
#         "series_id",
#         "series_label",
#         "year"
#     )
#     .agg(
#         sum("value").alias("annual_value")
#     )
# )

# window_spec = (
#     Window
#     .partitionBy("series_id")
#     .orderBy(
#         col("annual_value").desc(),
#         col("year").desc()
#     )
# )

# gold_bls_best_year_pyspark = (
#     annual_values
#     .withColumn(
#         "rn",
#         row_number().over(window_spec)
#     )
#     .filter(col("rn") == 1)
#     .select(
#         "series_id",
#         "series_label",
#         col("year").alias("best_year"),
#         "annual_value"
#     )
# )
```

```python
# Gold Layers
# question 3 - Annual Q01 value for PRS30006032 joined with population where available
# Spark SQL

@dp.materialized_view(
    name="bls.gold.gold_series_population",
    comment="Annual Q01 value for PRS30006032 joined with population where available"
)
def gold_series_population():

    return spark.sql("""
        SELECT
            b.year,
            b.series_id,
            b.series_label,
            b.period,
            b.value,
            p.population
        FROM bls.silver.silver_bls_data b
        LEFT JOIN bls.silver.silver_population p
            ON b.year = p.year
        WHERE b.series_id = 'PRS30006032'
          AND b.period = 'Q01'
    """)
    
# ============================================================
# Alternative implementation - PySpark
# ============================================================

# from pyspark.sql.functions import col

# bls_data = (
#     spark.read.table("bls.silver.silver_bls_data")
#     .filter(
#         (col("series_id") == "PRS30006032") &
#         (col("period") == "Q01")
#     )
# )

# population = (
#     spark.read.table("bls.silver.silver_population")
# )

# gold_series_population_pyspark = (
#     bls_data
#     .join(
#         population,
#         bls_data["year"] == population["year"],
#         "left"
#     )
#     .select(
#         bls_data["year"],
#         bls_data["series_id"],
#         bls_data["series_label"],
#         bls_data["period"],
#         bls_data["value"],
#         population["population"]
#     )
# )    
```
