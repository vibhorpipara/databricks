```python
%sql
CREATE CATALOG bls;
CREATE SCHEMA IF NOT EXISTS bls.bronze;
CREATE VOLUME IF NOT EXISTS bls.bronze.bls_raw;
CREATE SCHEMA IF NOT EXISTS bls.silver;
CREATE SCHEMA IF NOT EXISTS bls.gold;
```

```python
# creating one control table 
%sql
CREATE TABLE IF NOT EXISTS bls.bronze.bls_file_control (
    file_name STRING,
    source_url STRING,
    source_last_modified STRING,
    source_size BIGINT,
    content_hash STRING,
    volume_path STRING,
    ingestion_status STRING,
    first_ingested_at TIMESTAMP,
    last_checked_at TIMESTAMP
)
USING DELTA;
```
